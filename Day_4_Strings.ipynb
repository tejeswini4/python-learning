# 2. String Indexing

Each character in a string has a position number called an index.

Example:

String:

P   Y   T   H   O   N

Index:

0   1   2   3   4   5


Indexing starts from 0 in Python.

```python
word = "PYTHON"

print(word[0])
print(word[2])
print(word[5])
```

# Negative Indexing in Strings

Python also allows accessing characters from the end of a string.

Negative indexing starts with -1.

Example:

String:

P   Y   T   H   O   N

Index:

-6  -5  -4  -3  -2  -1


The last character always has index -1.

```python
word = "PYTHON"

print(word[-1])
print(word[-3])
print(word[-6])
```

# 3. String Slicing

String slicing is used to extract a part of a string.

Syntax:

string[start:end:step]


start:
- Starting index (included)

end:
- Ending index (excluded)

step:
- Number of jumps between characters


Example:

word = "PYTHON"

word[0:3]

Output:

PYT

```python
word = "PYTHON"

print(word[0:3])
print(word[2:5])
print(word[:4])
print(word[3:])
```

# Slicing with Step Value

Syntax:

string[start:end:step]


Step controls how many characters to skip.

Example:

word = "PYTHON"

word[0:6:2]

Characters selected:

P T O

```python
word = "PYTHON"

print(word[0:6:2])
print(word[::2])
print(word[::3])
```

# Reversing a String

We can reverse a string using slicing.

Syntax:

string[::-1]


The step value -1 moves from the end to the beginning.

```python
word = "PYTHON"

reverse = word[::-1]

print(reverse)
```

# Searching in Strings

Python provides methods to search for characters or words inside a string.

### find()

Returns the index of the first occurrence.

Syntax:

string.find("text")

If the text is not found, it returns **-1**.

---

### count()

Returns the number of times a character or word appears.

Syntax:

string.count("text")

```python
text = "Python Programming"

print(text.find("P"))
print(text.find("g"))
print(text.find("Java"))


print(text.count("m"))
print(text.count("Python"))
```

# Modifying Strings

Python provides methods to modify string contents.

### replace()

Replaces one part of a string with another.

Syntax:

string.replace(old, new)

---

### strip()

Removes spaces from the beginning and end of a string.

Syntax:

string.strip()

---

### split()

Splits a string into a list based on spaces (or another separator).

Syntax:

string.split()

```python
text = " Python Programming "

print(text.replace("Python", "AI"))

print(text.strip())

sentence = "Artifical Intelligence and Machine Learning"

print(sentence.split())
```

# Checking String Contents

Python provides methods to check the contents of a string.

### isalpha()

Returns True if all characters are alphabets.

---

### isdigit()

Returns True if all characters are digits.

---

### isalnum()

Returns True if all characters are alphabets or numbers.

---

### startswith()

Checks whether a string starts with a given value.

---

### endswith()

Checks whether a string ends with a given value.

```python
print("Python".isalpha())
print("12345".isdigit())
print("Python123".isalnum())
print("Python".startswith("Py"))
print("Python".endswith("on"))
```

# Check whether a string is a palindrome.

```python
text = input("Enter a string: ")

if text == text[::-1]:
  print("The string is a palindrome.")
else:
  print("The string is not a palindrome.")
```

```python
# using loops
text = input("Enter a string: ")

reverse = ""

for char in text:
    reverse = char + reverse

if text == reverse:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

# count the number of vowels in string

```python
text = input("Enter a string: ")

count = 0

for char in text:
  if char.lower() in "aeiou":
    count += 1

print("Number of vowels: ",count)
```

# count number of consonants in string

```python
text = input(" Enter a string :")

count = 0

for char in text:
  if char.isalpha() and  char.lower() not  in "aeiou":
    count += 1

print("Number of consonants: ",count)
```

# count the occurrence of a character

```python
text = input("Enter a string: ")
character = input("Enter a characer: ")

count = 0

for char in text:
  if char == character:
    count += 1

print("Number of occurrences: ",count)
```

# remove spaces from a string


```python
text = input("Enter a string: ")

new = ""

for char in text:
  if char != " ":
    new += char

print("String without spaces:",new)

```

# Convert Lowercase to Uppercase

```python
text = input("Enter a string: ")

new = ""

for char in text:
  if char.islower():
    new += char.upper()
  else:
    new += char

print("String with uppercase:",new)
```

# find duplicate characters in string

```python
text = input("Enter a string: ")

printed = ""

print("duplicate characters: ")

for char in text:
  if text.count(char) > 1 and char not in printed:
    print(char)
    printed += char
```
